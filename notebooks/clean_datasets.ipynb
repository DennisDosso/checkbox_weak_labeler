```python
import json

# Define your paths
json_path = r"C:\Users\dennis.dosso\Projects\OBJECT_DETECTION\datasets\signatures\signature_detection_coco\test\_annotations.merged.coco.json"
fixed_json_path =json_path
```

```python
# Load the JSON
with open(json_path, "r") as f:
    data = json.load(f)

# Loop through annotations and rename 'confidence' to 'score'
for ann in data.get("annotations", []):
    if "confidence" in ann:
        ann["score"] = ann.pop("confidence")

# Save the fixed JSON
with open(fixed_json_path, "w") as f:
    json.dump(data, f)
    
print("JSON successfully patched!")
```
